### Verificación de estar obteniendo los mismos vectores con la optimización

```python
import numpy as np
import pandas as pd
import json
```

```python
#Cargamos las rutas

df_Original = pd.read_csv("/files2/gerardo/KmerTopology/data/processed/selected_genomes/vectors_topB_escherichia.csv")
df_Optimizado = pd.read_csv("/files2/gerardo/KmerTopology/data/processed/optimized_pipeline/vectors_topB_escherichia.csv")
```

```python
# Cargamos las matrices para compararlas

vect_Original = df_Original.to_numpy()
vect_Optimizado = df_Optimizado.to_numpy()
```

```python
n = vect_Original.shape[0]
print("Numero de vectores",n)
```

```python
cont = 0
for i in range(n):
    if not np.array_equal(vect_Original[i], vect_Optimizado[i]) :
        print("Vector no.", i, "distinto")
    else:
        cont = cont+1
print(cont,"/",n,"vectores iguales")
```
