```python
# Obtenemos los nombres de nuestros 10 genomas de escherichia
ruta_csv = Path("../data/processed/escherichia_10.csv")
#Lo hacemos dataframe
df = pd.read_csv(ruta_csv)
```

```python
print(df)
```

```python
nombres = []
nombres2 = []
for nombre in df["name"]:
    if nombre.endswith(".fna"):
        nombres.append(nombre[:-4])
    else:
        nombres.append(nombre)

print(nombres)
```
